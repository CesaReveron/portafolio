# Análisis Exploratorio de Datos (EDA)
## Dataset Técnico - Métricas Operativas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

## 1. Carga de Datos

```python
df = pd.read_csv('../data/dataset_cleaned.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])
print(f'Dataset cargado: {len(df)} registros')
```

## 3. Distribución por Categorías

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

df['categoria'].value_counts().plot(kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Registros por Categoría')
axes[0].set_ylabel('Cantidad')

df['estado'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%')
axes[1].set_title('Distribución de Estados')

plt.tight_layout()
plt.savefig('../data/chart_category_status.png', dpi=150)
plt.show()
```

## 4. Análisis de Valor por Región

```python
fig, ax = plt.subplots(figsize=(10, 5))
df.groupby('region')['valor'].mean().sort_values().plot(kind='barh', color='coral')
ax.set_title('Valor Promedio por Región')
ax.set_xlabel('Valor Promedio')
plt.tight_layout()
plt.savefig('../data/chart_region_avg.png', dpi=150)
plt.show()
```

## 5. Tendencia Temporal

```python
fig, ax = plt.subplots(figsize=(14, 5))
df.set_index('timestamp').resample('ME')['valor'].mean().plot(ax=ax, marker='o', color='green')
ax.set_title('Tendencia del Valor Promedio en el Tiempo')
ax.set_ylabel('Valor Promedio')
ax.set_xlabel('Fecha')
plt.tight_layout()
plt.savefig('../data/chart_trend.png', dpi=150)
plt.show()
```

## 6. Boxplot por Estado

```python
fig, ax = plt.subplots(figsize=(10, 6))
df.boxplot(column='valor', by='estado', ax=ax)
ax.set_title('Distribución de Valor por Estado')
plt.suptitle('')
plt.tight_layout()
plt.savefig('../data/chart_boxplot.png', dpi=150)
plt.show()
```

## 7. KPIs Calculados

```python
kpis = {
    'Total Registros': len(df),
    'Valor Promedio': round(df['valor'].mean(), 2),
    'Valor Máximo': round(df['valor'].max(), 2),
    'Tasa de Errores (%)': round((df['estado'] == 'Error').mean() * 100, 2),
    'Tiempo Respuesta Promedio': round(df['tiempo_respuesta'].mean(), 2),
    'Mejor Región (por valor)': df.groupby('region')['valor'].mean().idxmax(),
    'Categoría con más errores': df[df['estado'] == 'Error']['categoria'].value_counts().idxmax()
}
for k, v in kpis.items():
    print(f'{k}: {v}')
```

## 8. Exportar Resultados

```python
kpis_df = pd.DataFrame(list(kpis.items()), columns=['Métrica', 'Valor'])
kpis_df.to_csv('../data/kpis_summary.csv', index=False)
print('KPIs exportados a ../data/kpis_summary.csv')
```
